```python
import sys
sys.path.append('./lib/Office365-REST-Python-Client')

from office365.sharepoint.client_context import ClientContext
from office365.runtime.auth.user_credential import UserCredential

test_client_credentials = UserCredential('bue@foryouandyourcustomers.com','ZiZO#z]1S[B\\+\\%x{+\\x5g')
test_team_site_url = 'https://bosnet.sharepoint.com/sites/Bossard_MasterDataSandbox'

ctx = ClientContext(test_team_site_url).with_credentials(test_client_credentials)
```

```python
entity_list = ctx.web.lists.get_by_title("Entities")
```

```python
def print_progress(items_read):
    print("Items read: {0}".format(items_read))
    
def get_total_count(target_list):
    """
    :type target_list: office365.sharepoint.lists.list.List
    """
    items = target_list.items.paged(False).top(1000)
    items.page_loaded += print_progress  # page load event
    result = items.get_items_count().execute_query()
    print("Total items count: {0}".format(result.value))
```

```python
get_total_count(entity_list)
```

```python
import json

ssot_file = '/Users/bue/projects/bossard/im/DB/Bossard.json'
with open(ssot_file, 'r') as src:
    ssot = json.load(src)
```

```python
ssot['_imprint_']
```

```python
def tr(element: dict, lang: str = 'en') -> str:
    if element is None or len(element) < 1:
        return ''
    return element.get(lang, '')
```

```python
for key, entity in ssot['entities'].items():
    item_to_create = entity_list.add_item({
        'Title': tr(entity['name']),
        'Description': tr(entity['descr']),
        'Synonyms': tr(entity['synonyms']),
        #'Name_DE': tr(entity['name'], 'de'),
        'Key': key,
    }).execute_query()
```
