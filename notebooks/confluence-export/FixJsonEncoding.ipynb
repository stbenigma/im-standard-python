```python
import re
# convert &#43; to unicode char
pattern = re.compile(r'&#([0-9]+);')
```

```python
match = re.search(pattern, "bla &#43; + 1")
match = pattern.search("bla &#43; + 1")
assert match is not None
match
```

```python
def fix_encoding(input) -> str:
    remainder = input
    match = pattern.search(remainder)
    while match is not None:
        codepoint_number = int(match.group(1))
        if codepoint_number == 34: # double quote
            repl = "\\\""
            print(f"Replace {match.group(0)} with {repl}")
        else:
            repl = chr(codepoint_number)
        #print(f"Replacing {matching_string} with {repl} from {match.group(0)}")
        remainder = remainder.replace(match.group(0), repl)
        match = pattern.search(remainder)
    return remainder
```

```python
input = "bla &#43; + 1\nbli &#232; flo ü"
result = fix_encoding(input)
print(f"-- Converted\n{input}\n-- to\n{result}")
```
